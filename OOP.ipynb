```python
# Inheritance in python
```

```python
class Animal:
    def __init__(self,name):
        self.name=name;
    def info(self):
        print("Animal name:",self.name);
class Dog(Animal):
    def sound(self):
        print(self.name,"barks");
d=Dog("Choummu");
d.info();
d.sound()
```

```python
# super() function in python -- it is used to call the parent class methods in
# childe or derived classes
```

```python
class Animal:
    def __init__(self,name):
        self.name=name;
    def info(self):
        print("Animal name:",self.name);
class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed;
    def details(self):
        print(self.name,"is a", self.breed ,"breed");
d=Dog("Choummu","desi(Sadak chap)");
d.info();
d.details()
```

```python
class Parent:
    def show(self):
        print("parent show");
class Child(Parent):
    def show(self):
        super().show()
        print("Child show");
c=Child();
c.show()
```

```python
# types of python inheritance 
```

```python
# 1) single inheritance
class Human:
    def __init__(self,name):
        self.name=name
class employee(Human):
    def show_role(self):
        print(self.name ,"is employee");
e1=employee("ritesh");
e1.show_role()
        
        
```

```python
# multiple inheritance 
class Human:
    def __init__(self,name):
        self.name=name;
class job:
    def __init__(self,salary):
        self.salary=salary
class me(Human,job):
    def __init__(self,name,salary):
        Human.__init__(self,name)
        job.__init__(self,salary)
    def details(self):
        print(self.name ,"and his salary is",self.salary);
p1=me("yash",340000)
p1.details()
    
```
