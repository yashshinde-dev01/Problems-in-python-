```python
# QUe 1)Write a Python program to count the frequency of words in a file.
```

```python
file=open('Example.txt','r')
content=file.read()
lst=content.split()
# print(content)
freq={};
for i in lst:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1;

print(freq)        
    
```

```python
# Que 7)WAP to write 1 to 20 numbers to file and read only odd numbers
```

```python
with open("num.txt",'w') as num:
    for i in range(1,21):
        num.write(str(i)+" ")
file=open("num.txt",'r')
s=file.read()
t=s.split()
for i in t:
    if int(i)%2!=0:
        print(i,end=" ")
```

```python
# Que 8)WAP to read palindrome string and number from file 
```

```python
with open("Pali_num.txt",'w') as pali_num:
    pali_num.write("aba\n 121 \n madam \n hello \n 422 \n # \n %"+" ")
op=open("Pali_num.txt",'r')
t=op.read()
arr=t.split()
print(arr)
for l in arr:
    if l==l[::-1]:
        print(l)
```

```python
# Que 9) WAP to read special symbols from file. 
```

```python
file=open("Pali_num.txt",'r')
read=file.read()
arr=read.split()
for ch in arr:
    if not ch.isalnum():
        print(ch)
```
