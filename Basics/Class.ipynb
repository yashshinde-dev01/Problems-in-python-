```python
# Que 3) use class name as vechile having attribute vechile type ,color,amd mileage inherit other classs which will display vechile display
```

```python
class Vechile:
    def __init__(self,vechType,color):
        self.vechType=vechType;
        self.color=color

class Car(Vechile):
    def __init__(self,vechType,color,milage):
        super().__init__(vechType,color)
        self.milage=milage

    def display(self):
        print(self.vechType,self.color,self.milage)

c1=Car("bike","red",123)
c1.display()
        
```

```python
# Que 4) ovverride the constructor using single level innheritance having product name,product type as a argument of first constructor and product 
# quantity argument of derived class constructor class, parent class constructor from child 
```

```python
class Parent:
    def __init__(self,pName,pType):
        self.pName=pName;
        self.pType=pType;

class child(Parent):
    def __init__(self,pName,pType,pqty):
        super().__init__(pName,pType)
        self.pqty=pqty;

    def display(self):
        print("product name:",self.pName)
        print("product Type:",self.pType)
        print("product quantity:",self.pqty)

p1=child("chappal",1,4)
p1.display()
        
        
```
