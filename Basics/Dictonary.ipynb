```python
# Que 9) Write a Python program to remove duplicates from the dictionary 
```

```python
d={1:'a',2:'b',3:'c',4:'d',5:'a',6:'d'}
d2={};
seen=set();
for key,val in d.items():
    if val not in seen:
        seen.add(val);
        d2[key]=val;
print(d2)        
```

```python
# Que 18). WAP to remove same element from list of values from dictionary. 
# 1:[11,12,13,11,5,12],2:[‘a’,’b’,’a’,’c’}
```

```python
d={1:[11,12,13,11,5,12],2:['a','b','a','c']}
lst=[];
for key in d:
    lst=[];
    for i in d[key]:
        if i not in lst:
            lst.append(i);
    d[key]=lst;
print(d)    
```
