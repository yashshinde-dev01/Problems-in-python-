```python
# Que 3) print number in sorted order without inbuilt function
```

```python
nums=[64, 34, 25, 12, 22, 11, 90];
for i in range(0,len(nums)):
    swapped=False;
    for j in range(0,len(nums)-1-i):
        if nums[j]>nums[j+1]:
            temp=nums[j];
            nums[j]=nums[j+1]
            nums[j+1]=temp;
            swapped=True;
    if swapped is False:
        break;
print(nums)            
```

```python
# Que 18) Write a Python program to find the first even and odd number in a given list of 
# numbers. Original list: [1, 3, 5, 7, 4, 1, 6, 8 
```

```python
# Que 24) Write a Python program that takes a list of integers and finds all pairs of integers that 
# differ by three. Return all pairs of integers in a list. 
# Sample Data: 
# ([0, 3, 4, 7, 9]) -> [[0, 3], [4, 7]] 
# [0, -3, -5, -7, -8] -> [[-3, 0], [-8, -5]] 
# ([1, 2, 3, 4, 5]) -> [[1, 4], [2, 5]] 
```

```python
l=[0, -3, -5, -7, -8]
l2=[];
for i in range(0,len(l)):
    for j in range(i+1,len(l)):
        diff=l[i]-l[j];
        if diff==3 or diff==-3:
            l2.append([l[i],l[j]]);
print(l2)            
        
```
