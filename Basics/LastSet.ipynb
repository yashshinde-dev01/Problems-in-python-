```python
# Que 26)Program to remove integers that are not repeated in value from the dictionary 
```

```python
d={'a':1,'b':2,'c':1,'d':2,'e':4,'f':'cat'}
lst=[v for v in d.values() if type(v)==int]

freq={}
for num in lst:
    if num in freq:
      freq[num]=freq.get(num,0)+1
    else:
        freq[num]=1
result={}
for k,v in d.items():
    if  type(v)==int and freq[v]>1:
        result[k]=v;
print(result)        
```

```python
# Que 42) Implement file handling program to count number of tabs, and to read special symbols from file.
```

```python
with open("three.txt", "w") as file:
    file.write("A\tB#@1")
with open('three.txt','r') as file:
    c=file.read()
no_tab=0
symcnt=0
for ch in c:
    if ch=='\t':
        no_tab+=1
    elif not ch.isalnum() and ch!='\n' and ch!=' ':
       symcnt+=1
print('total tab is :',no_tab)
print('total special symbol is:',symcnt)   
```
